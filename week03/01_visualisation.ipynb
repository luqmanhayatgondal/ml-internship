```python

import numpy as np
import matplotlib.pyplot as plt

# Chart 1: Monthly revenue
chart1_months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [41.2, 38.5, 46.9, 52.3, 49.8, 58.1]

# Chart 2: Incidents by type
incidents = {
    "Phishing": 128,
    "Credential stuffing": 96,
    "Malware": 74,
    "DDoS": 52,
    "Insider misuse": 31
}

# Chart 3: Response time distribution
rng = np.random.default_rng(42)
response_ms = rng.lognormal(mean=3.2, sigma=0.5, size=500)

# Chart 4: Ad spending and signups
rng = np.random.default_rng(7)
spend = rng.uniform(0, 100, 80)
signups = 2.5 * spend + rng.normal(0, 25, 80)

# Chart 5: Task times for three teams
rng = np.random.default_rng(3)
teams = [
    rng.normal(50, 8, 100),
    rng.normal(58, 12, 100),
    rng.normal(54, 5, 100)
]
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(chart1_months, revenue, marker="o", color="steelblue")

ax.set_title("Revenue Reaches a Six-Month High in June")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue ($ thousand)")

fig.tight_layout()
fig.savefig("figures/chart1_monthly_revenue.png", dpi=150)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

incident_names = list(incidents.keys())
incident_counts = list(incidents.values())

ax.barh(incident_names[::-1], incident_counts[::-1], color="teal")

ax.set_title("Phishing Causes the Most Security Incidents")
ax.set_xlabel("Number of Incidents (count)")
ax.set_ylabel("Incident Type")

fig.tight_layout()
fig.savefig("figures/chart2_incidents.png", dpi=150)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.hist(response_ms, bins=25, color="skyblue", edgecolor="black")

mean_value = np.mean(response_ms)
median_value = np.median(response_ms)

ax.axvline(mean_value, color="red", linestyle="--",
           label=f"Mean: {mean_value:.1f} ms")

ax.axvline(median_value, color="green", linestyle="-.",
           label=f"Median: {median_value:.1f} ms")

ax.set_title("Response Time Distribution")
ax.set_xlabel("Response Time (milliseconds)")
ax.set_ylabel("Frequency (count)")
ax.legend()

fig.tight_layout()
fig.savefig("figures/chart3_response_time.png", dpi=150)

plt.show()

print("Mean:", round(mean_value, 1), "ms")
print("Median:", round(median_value, 1), "ms")
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.scatter(spend, signups, color="steelblue", label="Actual Data")

slope, intercept = np.polyfit(spend, signups, 1)

fitted_signups = slope * spend + intercept

ax.plot(spend, fitted_signups, color="red",
        label="Fitted Line")

ax.set_title("Higher Ad Spending Is Associated with More Signups")
ax.set_xlabel("Ad Spending (units)")
ax.set_ylabel("Number of Signups (count)")
ax.legend()

fig.tight_layout()
fig.savefig("figures/chart4_spending_signups.png", dpi=150)

plt.show()

print("Slope:", round(slope, 2))
print("Intercept:", round(intercept, 2))
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(teams, tick_labels=["Team A", "Team B", "Team C"])

ax.set_title("Team B Has the Highest Median Task Time")
ax.set_xlabel("Teams")
ax.set_ylabel("Task Time (units)")

fig.tight_layout()
fig.savefig("figures/chart5_team_comparison.png", dpi=150)

plt.show()
```

## Task 2: Critique of Poor Charts

### Chart X: Customer Satisfaction

1. Problem: The y-axis starts at 96 instead of zero.
   Wrong conclusion: The difference between the two teams looks much larger than it really is.

2. Problem: The bars exaggerate the small difference between the scores.
   Wrong conclusion: A reader may think Team B is much better than Team A.

3. Problem: Red and green colours suggest good and bad performance.
   Wrong conclusion: A reader may judge the teams based on colour instead of actual scores.

4. Problem: The chart does not clearly show the exact score values.
   Wrong conclusion: Readers may misunderstand the actual difference between the teams.

### Chart Y: Share of Incident Types

1. Problem: The pie chart has nine slices.
   Wrong conclusion: Readers may find it difficult to compare all incident types.

2. Problem: It uses too many rainbow colours.
   Wrong conclusion: Readers may think the colours represent different levels of importance.

3. Problem: There are no labels or percentages.
   Wrong conclusion: Readers cannot easily identify the size of each incident type.

4. Problem: The slices are not arranged in a meaningful order.
   Wrong conclusion: Readers may struggle to identify the largest and smallest incident types.

```python
# Source: ChatGPT assistance; review and understand this code.

rng = np.random.default_rng(1)

incident_values = rng.integers(3, 14, 9)

incident_labels = [
    "Phishing",
    "Malware",
    "DDoS",
    "Ransomware",
    "Data Theft",
    "Brute Force",
    "Insider Threat",
    "Spam",
    "Other"
]

# Sort incidents from largest to smallest
sorted_indices = np.argsort(incident_values)

sorted_labels = np.array(incident_labels)[sorted_indices]
sorted_values = incident_values[sorted_indices]

fig, ax = plt.subplots(figsize=(9, 6))

ax.barh(sorted_labels, sorted_values, color="steelblue")

ax.set_title("Incident Types Ranked by Frequency")
ax.set_xlabel("Number of Incidents (count)")
ax.set_ylabel("Incident Type")

fig.tight_layout()
fig.savefig("figures/chartY_incident_types.png", dpi=150)

plt.show()
```

```python
# Source: ChatGPT assistance; review and understand this code.

import pandas as pd

# Load the cleaned Week 2 sales data
sales_df = pd.read_csv("../week02/data/processed/sales_clean.csv")

# Convert order_date into datetime format
sales_df["order_date"] = pd.to_datetime(sales_df["order_date"])

# Remove cancelled orders
sales_df = sales_df[sales_df["status"] != "Cancelled"].copy()

# Group revenue by month
monthly_revenue = (
    sales_df.groupby(sales_df["order_date"].dt.to_period("M"))["revenue"]
    .sum()
)

print("Monthly Revenue (Cancelled orders excluded):")
print(monthly_revenue)

print("\nTotal orders after excluding cancelled orders:", len(sales_df))
```

```python
# Source: ChatGPT assistance; review and understand this code.

import matplotlib.pyplot as plt

# Create Figure and Axes
fig, ax = plt.subplots(figsize=(9, 5))

# Draw the monthly revenue line chart
ax.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o",
    color="steelblue",
    linewidth=2
)

# Add title and axis labels
ax.set_title("Monthly Revenue Peaks in February 2026")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue ($)")

# Add grid for readability
ax.grid(axis="y", alpha=0.3)

# Save the chart
fig.tight_layout()
fig.savefig("figures/task3_monthly_sales_revenue.png", dpi=150)

plt.show()
```
