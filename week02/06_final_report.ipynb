# Week 2 - Final Sales Cleaning Report

## 1. Load Data and Apply Cleaning Function

This section loads the original sales dataset, applies the reusable cleaning function, and loads Wednesday's cleaned dataset for comparison.


```python
import pandas as pd
from sales_cleaning import clean_sales

# Load original raw sales data
df_raw = pd.read_csv("data/raw/sales.csv")

# Clean data using our reusable function
df_clean = clean_sales(df_raw)

# Load Wednesday's cleaned dataset
df_wednesday = pd.read_csv("data/processed/sales_clean.csv")

# Display dataset shapes
print("Raw dataset shape:", df_raw.shape)
print("Friday cleaned dataset shape:", df_clean.shape)
print("Wednesday cleaned dataset shape:", df_wednesday.shape)
```

## 2. Compare Friday and Wednesday Cleaning Results

This section compares both cleaned datasets to verify that the reusable cleaning function produces the same results as Wednesday's cleaning notebook.

```python
# Convert Wednesday dates into datetime format
df_wednesday["order_date"] = pd.to_datetime(
    df_wednesday["order_date"]
)

# Sort both datasets by order_id and reset indexes
df_friday_sorted = (
    df_clean.sort_values("order_id").reset_index(drop=True)
)

df_wednesday_sorted = (
    df_wednesday.sort_values("order_id").reset_index(drop=True)
)

# Compare both cleaned datasets
pd.testing.assert_frame_equal(
    df_friday_sorted,
    df_wednesday_sorted,
    check_dtype=False
)

print("SUCCESS: Friday and Wednesday cleaned datasets match!")
```

## 3. Final Data Validation

This section verifies that the cleaned dataset satisfies all assignment requirements.

```python
# Final validation checks
print("Final dataset shape:", df_clean.shape)
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Unique order IDs:", df_clean["order_id"].is_unique)
print("Unique regions:", df_clean["region"].nunique())
print("Unique statuses:", df_clean["status"].nunique())
print("All dates parsed:", df_clean["order_date"].notna().all())
print("Total revenue:", round(df_clean["revenue"].sum(), 2))

# Acceptance checks
assert len(df_clean) == 957
assert df_clean.shape[1] == 11
assert df_clean["order_id"].is_unique
assert df_clean["region"].nunique() == 4
assert df_clean["status"].nunique() == 3
assert df_clean["order_date"].notna().all()
assert df_clean.isna().sum().sum() == 0
assert df_clean.duplicated().sum() == 0
assert round(df_clean["revenue"].sum(), 2) == 4162140.15

print("\nAll final validation checks passed!")
```
