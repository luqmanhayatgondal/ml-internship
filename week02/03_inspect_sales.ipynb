```python

import pandas as pd

# Load the original sales data
df = pd.read_csv("data/raw/sales.csv")

# Q1: Number of rows and columns
print("Q1: Dataset Shape")
print(df.shape)

# Q2: Data types of each column
print("\nQ2: Data Types")
print(df.dtypes)

# Q3: Missing values in each column
print("\nQ3: Missing Values")
print(df.isnull().sum())

# Q4: Exact duplicate rows
print("\nQ4: Duplicate Rows")
print(df.duplicated().sum())

# Q5: Unique region labels
print("\nQ5: Unique Regions")
print("Count:", df["region"].nunique())
print(df["region"].unique())

# Q6: Unique status labels
print("\nQ6: Unique Statuses")
print("Count:", df["status"].nunique())
print(df["status"].unique())

# Q7: Distinct products, categories and customers
print("\nQ7: Distinct Values")
for column in ["product", "category", "customer"]:
    if column in df.columns:
        print(column, ":", df[column].nunique())

# Q8: Quantity statistics
print("\nQ8: Quantity Description")
print(df["quantity"].describe())

# Q9: Negative, zero and above 1000 quantities
print("\nQ9: Invalid Quantities")
print("Negative or zero:", (df["quantity"] <= 0).sum())
print("Above 1000:", (df["quantity"] > 1000).sum())

# Q10: Unit prices containing $
print("\nQ10: Prices Containing Dollar Sign")
print(df["unit_price"].astype(str).str.contains(r"\$", regex=True).sum())

# Q11: Different date formats
print("\nQ11: Date Format Patterns")
dates = df["order_date"].astype(str)

print("YYYY-MM-DD:", dates.str.contains(r"^\d{4}-\d{2}-\d{2}$").sum())
print("DD/MM/YYYY:", dates.str.contains(r"^\d{2}/\d{2}/\d{4}$").sum())
print("MM-DD-YYYY:", dates.str.contains(r"^\d{2}-\d{2}-\d{4}$").sum())
print("Other formats:", (
    ~dates.str.contains(
        r"^(\d{4}-\d{2}-\d{2}|\d{2}/\d{2}/\d{4}|\d{2}-\d{2}-\d{4})$",
        regex=True
    )
).sum())

# Q12: Is order_id unique?
print("\nQ12: Order ID Uniqueness")
print("Unique order IDs:", df["order_id"].is_unique)
print("Unique IDs count:", df["order_id"].nunique())
print("Total rows:", len(df))

# Final preview
print("\nFirst 5 Rows")
display(df.head())
```
