## Exercise 1: Create a 5x4 NumPy Array

This exercise creates a NumPy array containing numbers from 1 to 20 and reshapes it into 5 rows and 4 columns. It also displays the shape of the array.

```python
import numpy as np

numbers = np.arange(1, 21)

print(numbers)
matrix = numbers.reshape(5, 4)

print(matrix)

print(matrix.shape)
```

## Exercise 2: Array Properties

This exercise checks the array's shape, number of dimensions, total number of elements, and data type using NumPy attributes.

```python
print("Dimensions:", matrix.ndim)
print("Total elements:", matrix.size)
print("Data type:", matrix.dtype)
```

## Exercise 3: Extract the Middle 3x2 Block

This exercise uses NumPy slicing to extract the middle 3 rows and 2 columns from the original 5x4 matrix.

```python
middle = matrix[1:4, 1:3]

print(middle)
```

## Exercise 4: Boolean Filtering

This exercise filters the matrix to select only those values that are greater than 10 and divisible by 3 using NumPy boolean indexing.

```python
filtered = matrix[(matrix > 10) & (matrix % 3 == 0)]

print(filtered)
```

## Exercise 5: Replace Odd Numbers

This exercise uses NumPy's np.where() function to replace all odd numbers in the matrix with -1 while keeping even numbers unchanged.

```python
result = np.where(matrix % 2 != 0, -1, matrix)

print(result)
```

## Exercise 6: Row Sums and Column Means

This exercise calculates the sum of each row and the mean of each column using NumPy's sum() and mean() functions with the appropriate axis values.

```python
row_sum = matrix.sum(axis=1)
column_mean = matrix.mean(axis=0)

print("Row sums:", row_sum)
print("Column means:", column_mean)
```

## Exercise 7: Subtract Column Means Using Broadcasting

This exercise subtracts the mean of each column from its corresponding values using NumPy broadcasting. This centers each column around zero.

```python
column_means = matrix.mean(axis=0)

centered = matrix - column_means

print(centered)
```

## Exercise 8: Create a 9x9 Multiplication Table

This exercise uses NumPy broadcasting to multiply numbers from 1 to 9 and create a 9x9 multiplication table without using loops.

```python
a = np.arange(1, 10)
b = np.arange(1, 10)

table = a.reshape(9, 1) * b.reshape(1, 9)

print(table)
```

## Exercise 9: Random Normal Distribution

This exercise generates 1000 random numbers from a normal distribution with a mean of 50 and a standard deviation of 10. A fixed random seed of 42 ensures reproducible results. It also calculates the mean, standard deviation, minimum, and maximum values.

```python
rng = np.random.default_rng(42)

random_numbers = rng.normal(50, 10, 1000)

print("Mean:", random_numbers.mean())
print("Standard deviation:", random_numbers.std())
print("Minimum:", random_numbers.min())
print("Maximum:", random_numbers.max())
```

## Exercise 10: Demonstrate View vs Copy

This exercise demonstrates the difference between a NumPy view and a copy. Changes made to a view also affect the original array, while changes made to a copy do not affect the original array.

```python
demo_matrix = matrix.copy()

view = demo_matrix[0:2, 0:2]
copy_array = demo_matrix[0:2, 0:2].copy()

view[0, 0] = 999
copy_array[0, 1] = 888

print("Original demo matrix:")
print(demo_matrix)

print("View:")
print(view)

print("Copy:")
print(copy_array)
```
